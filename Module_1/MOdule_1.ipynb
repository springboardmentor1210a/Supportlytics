Checking for null values

```python
print(df.isnull().sum())
```

```python
df["Resolution_Date"] = df["Resolution_Date"].fillna("Not Resolved")
```

```python
df["Resolution_Code"] = df["Resolution_Code"].fillna("Pending")
```

```python
df["Customer_Satisfaction"] = df["Customer_Satisfaction"].fillna("0")
```

```python
df["Resolution_Notes"] = df["Resolution_Notes"].fillna("No Resolution Yet")
```

Tticket distribution

```python
print(df['Ticket_Type'].value_counts())
```

```python
import matplotlib.pyplot as plt
ticket_counts = df['Ticket_Type'].value_counts()

ticket_counts.plot(kind='bar')

plt.title("Ticket Type Distribution")

plt.xlabel("Ticket Type")

plt.ylabel("Number of Tickets")

plt.show()
```

```python
print(df['Priority'].value_counts())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_csv("./cleaned_customer_support_tickets_200k.csv")

# 1. Ticket Type Distribution

plt.figure(figsize=(7,5))
ax = sns.countplot(data=df, x="type", order=df["type"].value_counts().index, palette="Blues_d")
plt.title("Ticket Distribution by Type")
plt.xlabel("Ticket Type")
plt.ylabel("Count")
plt.xticks(rotation=30)
for container in ax.containers:
    ax.bar_label(container, fmt=lambda x: f"{int(x):,}")
plt.tight_layout()
plt.savefig("./images/M3/plot_ticket_type_distribution.png", dpi=200)
plt.show()

# 2. Top Issue Categories by Frequency

plt.figure(figsize=(8,5))
top_categories = df["category"].value_counts().head(10)
ax = sns.barplot(x=top_categories.values, y=top_categories.index, palette="viridis")
plt.title("Top 10 Issue Categories by Frequency")
plt.xlabel("Number of Tickets")
plt.ylabel("Category")
for container in ax.containers:
    ax.bar_label(container, fmt=lambda x: f"{int(x):,}")
plt.tight_layout()
plt.savefig("./images/M3/plot_top_categories.png", dpi=200)
plt.show()

# 3. Tickets by Priority

plt.figure(figsize=(7,5))
ax = sns.countplot(data=df, x="priority", order=df["priority"].value_counts().index, palette="Reds_d")
plt.title("Ticket Volume by Priority")
plt.xlabel("Priority")
plt.ylabel("Count")
for container in ax.containers:
    ax.bar_label(container, fmt=lambda x: f"{int(x):,}")
plt.tight_layout()
plt.savefig("./images/M3/plot_priority_distribution.png", dpi=200)
plt.show()

# 4. Tickets by Queue

plt.figure(figsize=(10,5))
ax = sns.countplot(data=df, x="queue", order=df["queue"].value_counts().index, palette="mako")
plt.title("Ticket Volume by Queue")
plt.xlabel("Queue")
plt.ylabel("Count")
for container in ax.containers:
    ax.bar_label(container, fmt=lambda x: f"{int(x):,}")
plt.tight_layout()
plt.savefig("./images/M3/plot_queue_distribution.png", dpi=200)
plt.show()
```

```python
import pandas as pd
df = pd.read_csv("./customer_support_tickets_200k.csv")
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df['ticket_created_date'] = pd.to_datetime(df['ticket_created_date'])
df['ticket_resolved_date'] = pd.to_datetime(df['ticket_resolved_date'])
df[['ticket_created_date', 'ticket_resolved_date']].dtypes
```

```python
df.dtypes
```

```python
df.info()
df.isnull().sum()
```

```python
df.drop(columns=['browser'], inplace=True)
print('browser' in df.columns)
print(df.shape)
```

```python
category_type_map = {
    'Account Suspension': 'Request',
    'Bug Report': 'Problem',
    'Data Sync Issue': 'Incident',
    'Feature Request': 'Request',
    'Login Issue': 'Incident',
    'Payment Problem': 'Problem',
    'Performance Issue': 'Problem',
    'Refund Request': 'Request',
    'Security Concern': 'Incident',
    'Subscription Cancellation': 'Request'
}
cat_idx = df.columns.get_loc('category')
df.insert(cat_idx + 1, 'type', df['category'].map(category_type_map))
df[['category', 'type']].head(10)
```

```python
print('--- Distribution by Type ---')
display(df['type'].value_counts().rename_axis('Type').reset_index(name='Count'))
print('--- Distribution by Priority ---')
display(df['priority'].value_counts().rename_axis('Priority').reset_index(name='Count'))
print('--- Distribution by Category ---')
display(df['category'].value_counts().rename_axis('Category').reset_index(name='Count'))
```

```python
df['resolution_duration'] = (df['ticket_resolved_date'] - df['ticket_created_date']).dt.days

priority_score_map = {'Low': 1, 'Medium': 2, 'High': 3, 'Urgent': 4}
df['priority_score'] = df['priority'].map(priority_score_map)

df[['priority', 'priority_score', 'ticket_created_date', 'ticket_resolved_date', 'resolution_duration']].head(10)
```

```python
df.duplicated().sum()
```

```python
queue_map = {
    'Account Suspension': 'Account Management',
    'Bug Report': 'Development Team',
    'Data Sync Issue': 'Technical Support',
    'Feature Request': 'Product Team',
    'Login Issue': 'Technical Support',
    'Payment Problem': 'Billing Support',
    'Performance Issue': 'Infrastructure Team',
    'Refund Request': 'Billing Support',
    'Security Concern': 'Security Team',
    'Subscription Cancellation': 'Account Management'
}
df['queue'] = df['category'].map(queue_map)
df[['category', 'queue']].head(10)
```

```python
df.to_csv('./cleaned_customer_support_tickets_200k.csv', index=False)
print('Exported successfully:', df.shape)
```

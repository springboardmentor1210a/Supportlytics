```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\Chrome\customer support ticket.csv")
```

```python
avg_resolution_priority = df.groupby("priority")["resolution_time_hours"].mean()

print(avg_resolution_priority)
```

```python
avg_resolution_priority.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Resolution Time by Priority")
plt.xlabel("Priority")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=0)
plt.show()
```

```python
avg_resolution_category = df.groupby("category")["resolution_time_hours"].mean()

print(avg_resolution_category)
```

```python
avg_resolution_category.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Resolution Time by Issue Category")
plt.xlabel("Category")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=45)
plt.show()
```

```python
region_performance = df.groupby("region")["resolution_time_hours"].mean().sort_values()

print(region_performance)
```

```python
region_performance.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Resolution Time by Region")
plt.xlabel("Region")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=45)
plt.show()
```

```python
df["status"].value_counts()
```

```python
high_priority_unresolved = df[
    (df["priority"].isin(["High", "Critical"])) &
    (df["status"].isin(["Open", "In Progress"]))
]

print(high_priority_unresolved.head())
```

```python
print("High-priority unresolved tickets:",
      len(high_priority_unresolved))
```

```python
high_priority_unresolved["priority"].value_counts().plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("High-Priority Unresolved Tickets")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
region_counts = df["region"].value_counts()

print(region_counts)
```

```python
plt.figure(figsize=(10,5))

plt.imshow(
    [region_counts.values],
    aspect="auto"
)

plt.colorbar(label="Number of Tickets")

plt.xticks(
    range(len(region_counts)),
    region_counts.index,
    rotation=45
)

plt.yticks([0], ["Tickets"])

plt.title("Ticket Concentration by Region")
plt.show()
```

```python
region_category = pd.crosstab(
    df["region"],
    df["category"]
)

print(region_category)
```

```python
region_category.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Issue Categories by Region")
plt.xlabel("Region")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.show()
```

```python
df["cluster"] = df["category"]
```

```python
cluster_size = df.groupby("cluster").size()
```

```python
cluster_performance = df.groupby("cluster")[
    "customer_satisfaction_score"
].mean()
```

```python
cluster_analysis = pd.DataFrame({
    "Cluster Size": cluster_size,
    "Average Performance Score": cluster_performance
})

print(cluster_analysis)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    cluster_analysis["Cluster Size"],
    cluster_analysis["Average Performance Score"]
)

for cluster in cluster_analysis.index:
    plt.annotate(
        cluster,
        (
            cluster_analysis.loc[cluster, "Cluster Size"],
            cluster_analysis.loc[cluster, "Average Performance Score"]
        )
    )

plt.title("Cluster Size vs Performance Score")
plt.xlabel("Cluster Size")
plt.ylabel("Average Customer Satisfaction Score")
plt.show()
```

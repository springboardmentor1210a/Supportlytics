# Module 2: Data Cleaning and Feature Engineering


```python
df.isnull().sum()
```

```python
text_columns = df.select_dtypes(include='object').columns

# Replace missing values
df[text_columns] = df[text_columns].fillna("Unknown")

print("Missing values in text columns handled successfully!")
```

```python
for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Extra spaces removed successfully!")
```

```python


df["Created_Date"] = pd.to_datetime(
    df["Created_Date"],
    errors="coerce"
)

df["Resolution_Date"] = pd.to_datetime(
    df["Resolution_Date"],
    errors="coerce"
)

print("Date columns converted successfully!")
```

```python
df["Resolution_Duration"] = (
    df["Resolution_Date"] - df["Created_Date"]
).dt.days

df[["Created_Date","Resolution_Date","Resolution_Duration"]].head()
```

```python
priority_map = {
    "Critical":4,
    "High":3,
    "Medium":2,
    "Low":1
}

df["Priority_Score"] = df["Priority"].map(priority_map)

df[["Priority","Priority_Score"]].head()
```

```python
df[["Resolution_Duration","Priority_Score"]].head(10)
```

```python
df.to_excel(
    "Supportlytics_Processed_Data.xlsx",
    index=False
)

print("Processed dataset saved successfully!")
```

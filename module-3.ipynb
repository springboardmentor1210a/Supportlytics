```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("IT_Support_Ticket_Dataset_Cleaned.csv")
print(df.head(2))
```

```python
print(df.describe(include='all'))
```

```python
status_counts = df["Status"].value_counts()

status_counts.plot(kind="bar")

plt.title("Ticket Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
priority_counts = df["Priority"].value_counts()

priority_counts.plot(kind="bar")

plt.title("Priority Distribution")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
ticket_counts = df["Ticket_Type"].value_counts()

ticket_counts.plot(kind="bar")

plt.title("Ticket Type Distribution")
plt.xlabel("Ticket Type")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
category_counts = df["Category"].value_counts()

category_counts.plot(kind="bar", figsize=(10,5))

plt.title("Category Distribution")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.hist(df["Resolution_Duration"].dropna(), bins=20)

plt.title("Resolution Duration Distribution")
plt.xlabel("Resolution Duration (Days)")
plt.ylabel("Number of Tickets")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.hist(df["Customer_Satisfaction"].dropna(), bins=5)

plt.title("Customer Satisfaction Distribution")
plt.xlabel("Customer Satisfaction")
plt.ylabel("Number of Tickets")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Count the number of tickets in each queue
queue_counts = df["Queue"].value_counts()

# Plot bar chart
plt.figure(figsize=(8,4))

queue_counts.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Tickets by Queue")
plt.xlabel("Queue")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

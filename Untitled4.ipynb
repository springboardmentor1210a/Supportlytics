```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\Chrome\customer support ticket.csv")
```

```python
average_resolution_time = df["resolution_time_hours"].mean()

print("Average Resolution Time:",
      round(average_resolution_time, 2),
      "hours")
```

```python
most_frequent_category = df["category"].value_counts().idxmax()

print("Most Frequent Category:",
      most_frequent_category)
```

```python
top_region = df.groupby("region")["customer_satisfaction_score"].mean().idxmax()

print("Top Performing Region:",
      top_region)
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Average Resolution Time",
        "Most Frequent Category",
        "Top Performing Region"
    ],
    "Value": [
        round(average_resolution_time, 2),
        most_frequent_category,
        top_region
    ]
})

kpi_summary
```

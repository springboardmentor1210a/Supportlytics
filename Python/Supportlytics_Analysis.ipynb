### Create new features: Resolution_Duration and Priority_Score

```python
# Resolution_Duration

import pandas as pd

df = pd.read_excel("Supportlytics_Complete_Dataset.xlsx")

# Convert columns to datetime
df["Created_Date"] = pd.to_datetime(df["Created_Date"])
df["Resolution_Date"] = pd.to_datetime(df["Resolution_Date"])

# Create Resolution_Duration in hours
df["Resolution_Duration"] = (
    df["Resolution_Date"] - df["Created_Date"]
).dt.total_seconds() / 3600
```

```python
## Priority_Score

priority_mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3,
    "Critical": 4
}

df["Priority_Score"] = df["Priority"].map(priority_mapping)
```

```python
df[[
    "Created_Date",
    "Resolution_Date",
    "Resolution_Duration",
    "Priority",
    "Priority_Score"
]].head()
```

```python
df.to_excel("Supportlytics_Feature_Engineered.xlsx", index=False)

print("Processed dataset saved successfully!")
```

### Module 6: Geographic and Category-Level Insight
#### Create Heatmaps of Ticket Concentration by Region Objective

```python
#### Count Tickets by Region
region_count = df.groupby('Region').size()

print(region_count)
```

```python
import matplotlib.pyplot as plt

heatmap_data = df.pivot_table(
    index='Region',
    columns='Priority',
    values='Ticket_ID',
    aggfunc='count',
    fill_value=0
)

plt.figure(figsize=(8,5))

plt.imshow(heatmap_data, aspect='auto', cmap='YlOrRd')

plt.colorbar(label='Ticket Count')
plt.xticks(range(len(heatmap_data.columns)), heatmap_data.columns)
plt.yticks(range(len(heatmap_data.index)), heatmap_data.index)

plt.title("Ticket Concentration Heatmap by Region")
plt.xlabel("Priority")
plt.ylabel("Region")

plt.show()
```

### Evaluate the Relationship Between Cluster Size and Performance Score

#### Calculate Cluster Size

```python
cluster_size = df.groupby('Cluster_Name').size()
```

#### Calculate Average Performance Score

```python
performance = df.groupby('Cluster_Name')['Customer_Satisfaction'].mean()
```

```python
cluster_performance = cluster_size.to_frame(name='Cluster_Size')

cluster_performance['Performance_Score'] = performance

print(cluster_performance)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(
    cluster_performance['Cluster_Size'],
    cluster_performance['Performance_Score']
)

plt.title("Cluster Size vs Performance Score")
plt.xlabel("Cluster Size")
plt.ylabel("Average Customer Satisfaction")

plt.grid(True)

plt.show()
```

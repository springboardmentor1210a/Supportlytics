```python
queue_map = {
    "account_access": "Account Management",
    "billing_problem": "Billing Support",
    "security_concern": "Security Team",
    "bug": "Development Team",
    "performance": "Infrastructure Team",
    "feature_request": "Development Team",
    "how_to": "Technical Support",
    "other": "Technical Support"
}

df["Queue"] = df["issue_type"].map(queue_map)
```

```python
df["Queue"].value_counts().plot(kind="bar", figsize=(8,5))

plt.title("Ticket Distribution by Queue")
plt.xlabel("Queue")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)

plt.show()
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(df["initial_message"])
```

```python
import numpy as np

df["Similarity_Score"] = np.asarray(tfidf_matrix.mean(axis=1)).ravel()
```

```python
df[["Similarity_Score"]].head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["category_encoded"] = encoder.fit_transform(df["category"])
df["issue_type_encoded"] = encoder.fit_transform(df["issue_type"])
df["priority_encoded"] = encoder.fit_transform(df["priority"])
```

```python
features = df[
    [
        "category_encoded",
        "issue_type_encoded",
        "priority_encoded",
        "Similarity_Score"
    ]
]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(scaled_features)
```

```python
df["Cluster"].value_counts()
```

```python
df["Cluster"].value_counts().sort_index().plot(kind="bar", figsize=(8,5))

plt.title("Cluster Distribution")
plt.xlabel("Cluster")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df.groupby("Cluster")["Similarity_Score"].mean()
```

```python
pd.crosstab(df["Cluster"], df["issue_type"]).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Cluster vs Issue Type")
plt.xlabel("Cluster")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
plt.figure(figsize=(8,6))

df.boxplot(
    column="resolution_time_hours",
    by="Cluster",
    showfliers=False
)

plt.title("Resolution Time by Cluster")
plt.suptitle("")
plt.xlabel("Cluster")
plt.ylabel("Resolution Time (Hours)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["resolution_time_hours"],
    df["csat_score"],
    c=df["Cluster"]
)

plt.xlabel("Resolution Time (Hours)")
plt.ylabel("CSAT Score")
plt.title("Resolution Time vs CSAT Score")
plt.show()
```

```python
sample_df = df.sample(n=1000, random_state=42)

plt.figure(figsize=(8,6))

plt.scatter(
    sample_df["resolution_time_hours"],
    sample_df["csat_score"],
    alpha=0.6
)

plt.xlabel("Resolution Time (Hours)")
plt.ylabel("CSAT Score")
plt.title("Resolution Time vs CSAT Score (Sample of 1000 Tickets)")
plt.grid(True)

plt.show()
```

```python

sentiment_mapping = {
    'very_negative': 1,
    'negative': 2,
    'neutral': 3,
    'positive': 4,
    'very_positive': 5
}

df['Performance_Score'] = df['customer_sentiment'].map(sentiment_mapping)


cluster_performance = df.groupby('cluster').agg(
    Cluster_Size=('cluster', 'count'),
    Avg_Performance_Score=('Performance_Score', 'mean')
).reset_index()

cluster_performance
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=cluster_performance,
    x='Cluster_Size',
    y='Avg_Performance_Score',
    s=250,
    color='#7E57C2'
)

for i, row in cluster_performance.iterrows():
    plt.text(
        row['Cluster_Size'] + 20,
        row['Avg_Performance_Score'],
        f"Cluster {int(row['cluster'])}",
        fontsize=9
    )

plt.title('Cluster Size vs Performance Score')
plt.xlabel('Cluster Size')
plt.ylabel('Average Performance Score')

plt.show()
```

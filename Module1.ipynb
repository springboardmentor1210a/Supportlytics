```python
# Remove duplicates
df = df.drop_duplicates()

print("Dataset Shape after Removing Duplicates:", df.shape)
```

```python
#Checking missing values
df.isnull().sum()
```

```python
#Handling missing data
text_columns = df.select_dtypes(include='object').columns
df[text_columns] = df[text_columns].fillna("Unknown")
```

```python
#create resolution duration
df['Resolution_Duration'] = df['resolution_time_hours']
```

```python
#create priority score
priority_score = {
    'Low': 1,
    'Medium': 2,
    'High': 3,
    'Urgent': 4
}

df['Priority_Score'] = df['priority'].map(priority_score)
```

```python
#Verify new columns
df[['priority', 'Priority_Score', 'Resolution_Duration']].head()
```

```python
#cleaned dataset
df.to_csv("cleaned_customer_support_tickets.csv", index=False)
print("Cleaned dataset saved successfully!")
```

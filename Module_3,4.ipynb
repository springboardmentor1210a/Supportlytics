```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/customer_support_tickets_200k.csv')
plt.figure(figsize=(8,5))
df['category'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Ticket Category Distribution")
plt.xlabel("Ticket Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['category'].value_counts().head(10).plot(kind='bar', color='orange')
plt.title("Top Categories by Frequency")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
df['customer_segment'].value_counts().head(10).plot(kind='bar', color='green')
plt.title("Top Customer Segments by Frequency")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7,5))
df['priority'].value_counts().plot(kind='bar', color='red')
plt.title("Ticket Distribution by Priority")
plt.xlabel("Priority")
plt.ylabel("Count")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
df['channel'].value_counts().head(10).plot(kind='bar', color='purple')
plt.title("Tickets by Channel")
plt.xlabel("Channel")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
cluster_distribution = df['customer_segment'].value_counts().head(10).sort_values(ascending=False)
plt.figure(figsize=(10,5))
cluster_distribution.plot(kind='bar', color='teal')
plt.title("Distribution of Customer Segments")
plt.xlabel("Customer Segment")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.countplot(data=df,
              x='customer_segment',
              hue='category')
plt.xticks(rotation=90)
plt.title("Customer Segment vs Ticket Category")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.boxplot(data=df,
            x='priority',
            y='customer_satisfaction_score')
plt.title("Customer Satisfaction Score by Priority")
plt.xlabel("Priority")
plt.ylabel("Customer Satisfaction Score")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.scatterplot(data=df,
                x='first_response_time_hours',
                y='resolution_time_hours',
                hue='priority')

plt.title("First Response Time vs Resolution Time by Priority")
plt.xlabel("First Response Time (hours)")
plt.ylabel("Resolution Time (hours)")
plt.show()
```

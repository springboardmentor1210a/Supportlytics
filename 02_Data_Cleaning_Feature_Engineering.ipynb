# Module 2: Data Cleaning and Feature Engineering

## Objective

This notebook cleans the IT Support Ticket dataset by handling missing values, standardizing text fields, removing duplicates, creating new features, and saving the processed dataset for further analysis.

```python
import pandas as pd
import numpy as np
import os
import warnings

warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../data/IT_Support_Tickets.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum()/len(df))*100

missing_df = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage.round(2)
})

missing_df
```

```python
text_columns = [
    "Description",
    "Tags",
    "Product",
    "Resolution_Notes"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")
```

```python
df[text_columns].isnull().sum()
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df.head()
```

```python
for col in text_columns:
    df[col] = df[col].str.title()
```

```python
for col in text_columns:
    df[col] = df[col].replace("", "Unknown")
```

```python
df.head()
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
print(df["Ticket_ID"].duplicated().sum())
```

```python
df["Created_Date"] = pd.to_datetime(df["Created_Date"])

df["Resolution_Date"] = pd.to_datetime(df["Resolution_Date"])


df["First_Response_Date"] = pd.to_datetime(df["First_Response_Date"])
```

```python
df.info()
```

```python
df["Resolution_Duration"] = (
    df["Resolution_Date"] -
    df["Created_Date"]
).dt.total_seconds()/3600
```

```python
df[
[
"Created_Date",
"Resolution_Date",
"Resolution_Duration"
]
].head()
```

```python
priority_map = {

"Low":1,

"Medium":2,

"High":3,

"Critical":4

}

df["Priority_Score"] = df["Priority"].map(priority_map)
```

```python
df[
[
"Priority",
"Priority_Score"
]
].head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.head()
```

```python
os.makedirs("../data/processed",exist_ok=True)
```

```python
df.to_csv(
"../data/processed/cleaned_IT_Support_Tickets.csv",
index=False
)

print("Dataset saved successfully.")
```

# Feature Engineering Summary

## Resolution_Duration

Formula:

Resolution_Date − Created_Date

Unit:

Hours

Purpose:

Measures the total time required to resolve a ticket.

Business Value:

Helps evaluate support team performance.

---

## Priority_Score

Mapping

Low = 1

Medium = 2

High = 3

Critical = 4

Purpose

Converts priority into numerical values for analytics and machine learning.

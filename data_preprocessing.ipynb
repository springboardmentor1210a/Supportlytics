```python
# Ticket distribution by Priority
priority_distribution = (
    df['priority']
    .value_counts()
    .rename_axis('Priority')
    .reset_index(name='Ticket Count')
)

priority_distribution
```

```python
# Fill missing browser values
df['browser'] = df['browser'].fillna('Unknown')
```

```python
print(df['browser'].isnull().sum())
```

```python
text_columns = [
    'customer_name',
    'customer_email',
    'product',
    'category',
    'issue_description',
    'resolution_notes',
    'priority',
    'status',
    'channel',
    'region',
    'customer_gender',
    'subscription_type',
    'operating_system',
    'browser',
    'payment_method',
    'language',
    'preferred_contact_time',
    'customer_segment'
]

df[text_columns] = df[text_columns].apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)
```

```python
print("Missing Browser Values:", df['browser'].isnull().sum())

print("\nUnique Browser Values:")
print(df['browser'].unique())
```

```python
# Convert date columns to datetime format
df['ticket_created_date'] = pd.to_datetime(
    df['ticket_created_date'],
    errors='coerce'
)

df['ticket_resolved_date'] = pd.to_datetime(
    df['ticket_resolved_date'],
    errors='coerce'
)
```

```python
print(df[['ticket_created_date', 'ticket_resolved_date']].dtypes)
```

```python
# Create Resolution_Duration in days
df['Resolution_Duration'] = (
    df['ticket_resolved_date'] -
    df['ticket_created_date']
).dt.days
```

```python
df[['ticket_created_date',
    'ticket_resolved_date',
    'Resolution_Duration']].head()
```

```python
df['Resolution_Duration_Hours'] = (
    df['ticket_resolved_date'] -
    df['ticket_created_date']
).dt.total_seconds() / 3600
```

```python
priority_mapping = {
    'Low': 1,
    'Medium': 2,
    'High': 3,
    'Urgent': 4
}

df['Priority_Score'] = df['priority'].map(priority_mapping)
```

```python
df[['priority', 'Priority_Score']].head(10)
```

```python
print(df[['Resolution_Duration', 'Priority_Score']].head())
```

```python
print("Missing Resolution_Duration:",
      df['Resolution_Duration'].isnull().sum())

print("Missing Priority_Score:",
      df['Priority_Score'].isnull().sum())
```

```python
# Map Category to Ticket Type
type_mapping = {
    'Feature Request': 'Request',
    'Refund Request': 'Request',
    'Subscription Cancellation': 'Request',

    'Login Issue': 'Incident',
    'Account Suspension': 'Incident',
    'Payment Problem': 'Incident',

    'Performance Issue': 'Problem',
    'Bug Report': 'Problem',
    'Data Sync Issue': 'Problem',
    'Security Concern': 'Problem'
}

df['type'] = df['category'].map(type_mapping)
```

```python
df[['category', 'type']].head(15)
```

```python
type_distribution = (
    df['type']
      .value_counts()
      .rename_axis('Type')
      .reset_index(name='Ticket Count')
)

type_distribution['Percentage (%)'] = (
    type_distribution['Ticket Count']
    / len(df)
    * 100
).round(2)

type_distribution
```

```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\Chrome\customer support ticket.csv")
```

```python
import matplotlib.pyplot as plt

df["status"].value_counts().plot(
    kind="bar",
    figsize=(6,4),
    color="skyblue"
)

plt.title("Ticket Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df["category"].value_counts().plot(
    kind="bar",
    figsize=(6,4),
    color="green"
)

plt.title("Ticket Category Distribution")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df["priority"].value_counts().plot(
    kind="bar",
    figsize=(6,4),
    color="red"
)

plt.title("Ticket Priority Distribution")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df.columns
```

```python
def assign_queue(priority):
    if priority == "Critical":
        return "Queue 1"
    elif priority == "High":
        return "Queue 2"
    elif priority == "Medium":
        return "Queue 3"
    else:
        return "Queue 4"

df["queue"] = df["priority"].apply(assign_queue)
```

```python
df[["priority", "queue"]].head()
```

```python
import matplotlib.pyplot as plt

df["queue"].value_counts().plot(
    kind="bar",
    figsize=(6,4),
    color="purple"
)

plt.title("Ticket Distribution by Queue")
plt.xlabel("Queue")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
df["similarity_score"] = df["issue_complexity_score"] / 10
```

```python
df["cluster"] = df["category"]
```

```python
cluster_similarity = df.groupby("cluster")["similarity_score"].mean()

print(cluster_similarity)
```

```python
import matplotlib.pyplot as plt

cluster_similarity.plot(
    kind="bar",
    figsize=(8,5),
    color="green"
)

plt.title("Average Similarity Score by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Average Similarity Score")
plt.show()
```

```python
cluster_size = df.groupby(["cluster","category"]).size().unstack()

cluster_size.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Cluster Size vs Issue Type")
plt.xlabel("Cluster")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
plt.figure(figsize=(8,5))

df.boxplot(column="resolution_time_hours", by="priority")

plt.title("Resolution Time by Priority")
plt.suptitle("")
plt.xlabel("Priority")
plt.ylabel("Resolution Time (Hours)")
plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(
    df["issue_complexity_score"],
    df["resolution_time_hours"],
    alpha=0.5
)

plt.title("Complexity vs Resolution Time")
plt.xlabel("Issue Complexity Score")
plt.ylabel("Resolution Time (Hours)")
plt.show()
```
